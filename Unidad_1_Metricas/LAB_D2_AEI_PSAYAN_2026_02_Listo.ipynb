## Marco conceptual — Del modelo DELTA al modelo DELTA Plus

El modelo DELTA fue formulado por Davenport, Harris y Morison en *Analytics at Work* (2010) como un instrumento para evaluar la madurez analítica de una organización a partir de cinco dimensiones: **Data**, referida a la disponibilidad, integración y calidad de los datos; **Enterprise**, que examina si la capacidad analítica se encuentra dispersa en áreas aisladas o coordinada a escala institucional; **Leadership**, que valora el compromiso efectivo de la dirección con la decisión basada en evidencia; **Targets**, que analiza si los esfuerzos analíticos se concentran en los objetivos de mayor valor para el negocio; y **Analysts**, que considera la disponibilidad de profesionales con las competencias requeridas. El principio central del modelo establece que estas dimensiones deben progresar de manera conjunta, dado que un avance aislado en una de ellas no eleva la madurez global de la organización.

Con posterioridad, y ante la transformación del entorno tecnológico de la analítica empresarial, el modelo fue ampliado bajo la denominación **DELTA Plus**, que incorpora dos dimensiones adicionales. La primera es **Technology**, que comprende la infraestructura y las plataformas que permiten desarrollar y desplegar capacidades analíticas: soluciones en la nube, *data warehouses*, *data lakes* y arquitecturas *lakehouse*, entornos de procesamiento de grandes volúmenes de información, herramientas de visualización, plataformas analíticas corporativas, componentes de código abierto e infraestructura para el desarrollo de modelos de aprendizaje automático e inteligencia artificial. Su incorporación obedece a una constatación de orden práctico: una organización puede disponer de datos pertinentes y de analistas calificados y presentar, aun así, una madurez analítica limitada si carece de la tecnología necesaria para almacenar, procesar, analizar y poner en uso esa información de manera periódica y reproducible.

La segunda dimensión añadida es **Analytical Techniques**, que evalúa la variedad y la sofisticación de las técnicas empleadas para convertir los datos en información útil para la decisión. Estas técnicas admiten una progresión reconocible: la analítica **descriptiva** establece qué ocurrió; la **diagnóstica** indaga por qué ocurrió; la **predictiva** estima qué podría ocurrir; y la **prescriptiva** determina qué conviene hacer. Sobre esa base se incorporan progresivamente el aprendizaje automático, las técnicas de segmentación, los modelos de clasificación, los pronósticos, los métodos de optimización y las aplicaciones de inteligencia artificial. Dichas técnicas se presentan aquí como ilustración de la evolución de la capacidad analítica de una organización y serán objeto de desarrollo en semanas posteriores del curso; el presente laboratorio se sitúa en los niveles descriptivo y diagnóstico.

## Paso 2: Descarga de información financiera real desde la SMV

Se emplea el servicio web oficial de la SMV para consultar las principales cuentas y el Estado de Situación Financiera. Las operaciones utilizadas son `obtener_InfoFinanciera` y `obtener_BalanceGeneral`. El nombre oficial `obtener_GanciaPerdida` también se conserva para consultas del Estado de Resultados.


```python
SERVICIO_SMV = "https://mvnet.smv.gob.pe/ws_od_eeff/WebServiceInfoFinanciera.asmx"
EJERCICIO = 2024
PERIODO = "A"   # Anual
TIPO = "I"      # Individual
AZUCARERAS = ["CASA GRANDE", "CARTAVIO", "AGROINDUSTRIAL LAREDO", "POMALCA"]

def etiqueta_local(etiqueta):
    '''Retira el espacio de nombres de una etiqueta XML.'''
    return str(etiqueta).split("}")[-1]

def payload_a_dataframe(payload):
    '''Convierte el contenido JSON o XML incluido en la respuesta SOAP en DataFrame.'''
    payload = html.unescape(str(payload)).strip()
    try:
        objeto = json.loads(payload)
        if isinstance(objeto, dict):
            objeto = objeto.get("rows", objeto.get("data", objeto))
        if isinstance(objeto, dict):
            objeto = [objeto]
        return pd.DataFrame(objeto)
    except json.JSONDecodeError:
        raiz = ET.fromstring(payload)
        registros = []
        for nodo in raiz.iter():
            hijos = list(nodo)
            if len(hijos) >= 5 and all(not list(hijo) for hijo in hijos):
                registros.append({etiqueta_local(h.tag): h.text for h in hijos})
        return pd.DataFrame(registros)

def descargar_smv(operacion, ejercicio=EJERCICIO, periodo=PERIODO, tipo=TIPO):
    '''Consulta una operación oficial del servicio SOAP de la SMV.'''
    sobre = f'''<?xml version="1.0" encoding="utf-8"?>
    <soap:Envelope xmlns:soap="http://schemas.xmlsoap.org/soap/envelope/">
      <soap:Body><{operacion} xmlns="http://tempuri.org/">
        <Ejercicio>{ejercicio}</Ejercicio><Periodo>{periodo}</Periodo><Tipo>{tipo}</Tipo>
      </{operacion}></soap:Body>
    </soap:Envelope>'''
    respuesta = requests.post(
        SERVICIO_SMV,
        data=sobre.encode("utf-8"),
        timeout=90,
        headers={
            "Content-Type": "text/xml; charset=utf-8",
            "SOAPAction": f'"http://tempuri.org/{operacion}"',
        },
    )
    respuesta.raise_for_status()
    raiz = ET.fromstring(respuesta.content)
    resultado = next(
        (n for n in raiz.iter() if etiqueta_local(n.tag) == f"{operacion}Result"),
        None,
    )
    if resultado is None or not (resultado.text or "").strip():
        raise ValueError(f"La SMV no devolvió datos para {operacion}.")
    datos = payload_a_dataframe(resultado.text)
    if datos.empty:
        raise ValueError(f"La respuesta de {operacion} no contiene registros.")
    return datos

principales = descargar_smv("obtener_InfoFinanciera")
balance = descargar_smv("obtener_BalanceGeneral")

print("Principales cuentas:", principales.shape)
print("Estado de situación financiera:", balance.shape)
display(principales.head(3))
```

## Paso 3: Normalización, filtrado y validación del sector

Se normalizan los nombres de las columnas, se filtran las cuatro empresas y se verifica que todas estén representadas. No se utiliza una base simulada ni un respaldo inventado.


```python
def nombre_columna(texto):
    texto = re.sub(r"(?<!^)(?=[A-Z])", "_", str(texto))
    texto = unicodedata.normalize("NFKD", texto).encode("ascii", "ignore").decode()
    return re.sub(r"[^a-z0-9]+", "_", texto.lower()).strip("_")

def preparar(df):
    df = df.copy()
    df.columns = [nombre_columna(c) for c in df.columns]
    return df

def resolver_columna(df, candidatos):
    '''Encuentra una columna aun cuando la SMV cambie mayúsculas o convenciones de nombre.'''
    disponibles = {nombre_columna(c): c for c in df.columns}
    for candidato in candidatos:
        clave = nombre_columna(candidato)
        if clave in disponibles:
            return disponibles[clave]
    raise KeyError(f"No se encontró ninguna de estas columnas: {candidatos}. Disponibles: {list(df.columns)}")

def estandarizar_empresa(df):
    df = preparar(df)
    origen = resolver_columna(df, ["nombre_empresa", "razon_social", "empresa"])
    return df.rename(columns={origen: "nombre_empresa"})

def filtrar_azucareras(df):
    patron = "|".join(map(re.escape, AZUCARERAS))
    return df[
        df["nombre_empresa"].astype(str).str.upper().str.contains(patron, regex=True, na=False)
    ].copy()

principales = filtrar_azucareras(estandarizar_empresa(principales))
balance = filtrar_azucareras(estandarizar_empresa(balance))

empresas = sorted(principales["nombre_empresa"].dropna().unique())
print("Empresas encontradas:")
for empresa in empresas:
    print("-", empresa)

faltantes = [nombre for nombre in AZUCARERAS if not any(nombre in e.upper() for e in empresas)]
if faltantes:
    raise ValueError(f"No se localizaron todas las empresas. Faltantes: {faltantes}")
```

## Paso 4: Construcción reproducible de los indicadores financieros

| Familia | Indicador | Fórmula |
|---|---|---|
| Rentabilidad | Margen neto | Utilidad neta / Ingresos |
| Rentabilidad | ROA | Utilidad neta / Activo total |
| Rentabilidad | ROE | Utilidad neta / Patrimonio |
| Liquidez | Razón corriente | Activo corriente / Pasivo corriente |
| Endeudamiento | Endeudamiento | Pasivo total / Activo total |
| Endeudamiento | Apalancamiento | Pasivo total / Patrimonio |

La razón corriente se obtiene del Estado de Situación Financiera mediante las descripciones oficiales de activos y pasivos corrientes.


```python
def renombrar_primera(df, destino, candidatos):
    if destino in df.columns:
        return df
    return df.rename(columns={resolver_columna(df, candidatos): destino})

# Estandarización de las cuentas principales que devuelve obtener_InfoFinanciera.
for destino, candidatos in {
    "ventas": ["total_ingreso", "total_ingresos", "ingresos", "ventas_netas"],
    "utilidad_neta": ["utilidad_neta", "ganancia_perdida_neta", "resultado_neto"],
    "activo_total": ["activo_total", "total_activo", "total_activos"],
    "pasivo_total": ["pasivo_total", "total_pasivo", "total_pasivos"],
    "patrimonio": ["patrimonio_total", "total_patrimonio", "patrimonio"],
}.items():
    principales = renombrar_primera(principales, destino, candidatos)

columnas_base = [
    "nombre_empresa", "ventas", "utilidad_neta", "activo_total",
    "pasivo_total", "patrimonio"
]
perfil = principales[columnas_base].drop_duplicates("nombre_empresa").copy()
perfil = perfil.rename(columns={"nombre_empresa": "empresa"})

def a_numero(valor):
    '''Convierte importes de la SMV aunque incluyan separadores, espacios o paréntesis.'''
    if pd.isna(valor):
        return np.nan
    if isinstance(valor, (int, float, np.number)):
        return float(valor)
    texto = str(valor).strip().replace("\u00a0", "").replace(" ", "")
    negativo = texto.startswith("(") and texto.endswith(")")
    texto = texto.strip("()")
    # La SMV normalmente usa punto decimal. Las comas se eliminan como separador de miles.
    texto = texto.replace(",", "")
    texto = re.sub(r"[^0-9.\-]", "", texto)
    if texto in {"", "-", "."}:
        return np.nan
    numero = float(texto)
    return -numero if negativo else numero

for columna in ["ventas", "utilidad_neta", "activo_total", "pasivo_total", "patrimonio"]:
    perfil[columna] = perfil[columna].map(a_numero)

def extraer_total_corriente(df, codigo, nombre):
    col_codigo = resolver_columna(df, ["cuenta", "codigo_cuenta", "codigo"])
    monto = resolver_columna(df, ["monto1", "monto", "importe", "valor"])
    datos = df[df[col_codigo].astype(str).str.strip() == codigo][
        ["nombre_empresa", monto]].copy()
    datos[monto] = datos[monto].map(a_numero)
    datos = datos.dropna().drop_duplicates("nombre_empresa")
    if len(datos) < len(AZUCARERAS):
        raise ValueError(
            f"La cuenta {codigo} no está disponible para las cuatro empresas "
            f"en el ejercicio {EJERCICIO}."
        )
    return datos.rename(columns={"nombre_empresa": "empresa", monto: nombre})

perfil = perfil.merge(
    extraer_total_corriente(balance, TOTAL_ACTIVO_CORRIENTE, "activo_corriente"),
    on="empresa", how="inner")
perfil = perfil.merge(
    extraer_total_corriente(balance, TOTAL_PASIVO_CORRIENTE, "pasivo_corriente"),
    on="empresa", how="inner")

if len(perfil) < 4:
    raise ValueError("No fue posible construir el perfil completo de las cuatro empresas.")

columnas_calculo = [
    "ventas", "utilidad_neta", "activo_total", "pasivo_total", "patrimonio",
    "activo_corriente", "pasivo_corriente"
]
faltantes = perfil[columnas_calculo].isna().sum()
if faltantes.any():
    raise ValueError(
        "Existen cuentas sin valor numérico: " +
        ", ".join(f"{c} ({n})" for c, n in faltantes.items() if n)
    )

if (perfil[["ventas", "activo_total", "patrimonio", "pasivo_corriente"]] == 0).any().any():
    raise ValueError("Existe un denominador igual a cero; revise las cuentas antes de calcular.")

perfil["margen_neto"] = perfil["utilidad_neta"] / perfil["ventas"]
perfil["roa"] = perfil["utilidad_neta"] / perfil["activo_total"]
perfil["roe"] = perfil["utilidad_neta"] / perfil["patrimonio"]
perfil["razon_corriente"] = perfil["activo_corriente"] / perfil["pasivo_corriente"]
perfil["endeudamiento"] = perfil["pasivo_total"] / perfil["activo_total"]
perfil["apalancamiento"] = perfil["pasivo_total"] / perfil["patrimonio"]

numeros = perfil.select_dtypes(include="number").to_numpy()
if np.isinf(numeros[~np.isnan(numeros)]).any():
    raise ValueError("Se obtuvo una división entre cero; revise las cuentas de la empresa afectada.")

display(perfil.round(3))
```

**Pregunta 1.** Identifique la empresa con mayor ROE y la empresa con mayor endeudamiento. ¿Coinciden? Interprete la relación observada sin afirmar causalidad.

**Respuesta:** Con los datos completos de las cuatro empresas, **Cartavio** presenta el mayor ROE (24.0%) y también el mayor nivel de endeudamiento (58.6% del activo). Por lo tanto, **sí coinciden**.

La relación es aritmética, no causal: el ROE divide la utilidad neta entre el patrimonio. Al tener un mayor endeudamiento, el patrimonio de Cartavio es proporcionalmente menor frente a su pasivo (apalancamiento de 1.42 veces, contra 0.53 de Laredo o 0.83 de Casa Grande). Con una base patrimonial más pequeña, la utilidad produce un ROE mayor — no porque el endeudamiento "genere" rentabilidad, sino porque reduce el denominador sobre el que esa rentabilidad se mide.

El caso de **Pomalca** confirma la ausencia de causalidad: a pesar de tener un alto endeudamiento (50.2%), su ROE es el más bajo del grupo (1.4%), lo que demuestra que el apalancamiento financiero solo magnifica el indicador si la empresa genera suficiente utilidad neta. Un ROE alto apalancado en deuda no es evidencia, por sí solo, de mejor gestión operativa.

### Ejercicio 4 — Consulta empresarial con condiciones múltiples en DuckDB

En la Actividad 2 se empleó DuckDB para reproducir un ordenamiento ya obtenido con Polars, con la finalidad de comprobar la reproducibilidad del resultado entre dos motores distintos. Una consulta de negocio, sin embargo, no se limita a ordenar: delimita un subconjunto de la población mediante criterios definidos previamente por la gerencia.

En este ejercicio deberá construir una consulta que identifique aquellas empresas que satisfacen **simultáneamente** dos condiciones financieras —capacidad de atender obligaciones de corto plazo y generación de rentabilidad positiva para el accionista— y que presente el resultado ordenado según su exposición al endeudamiento. Este es el procedimiento habitual mediante el cual un área de riesgos delimita una cartera de empresas elegibles.

**Se solicita:**

1. Construir sobre `perfil_reto` una consulta SQL que incluya `SELECT`, `WHERE`, condiciones múltiples enlazadas con `AND` y `ORDER BY`.
2. Establecer como criterios una razón corriente mayor o igual a 1 y un ROE positivo.
3. Ordenar el resultado por endeudamiento, de menor a mayor.
4. Determinar cuántas empresas satisfacen ambos criterios.

**Tiempo estimado:** 10 a 12 minutos.


```python
CONSULTA_RETO = """
SELECT empresa,
       ROUND(razon_corriente, 2)     AS razon_corriente,
       ROUND(roe * 100, 1)           AS roe_pct,
       ROUND(endeudamiento * 100, 1) AS endeudamiento_pct
FROM perfil_reto
WHERE razon_corriente >= 1
  AND roe > 0
ORDER BY endeudamiento_pct ASC
"""

resultado_reto = duckdb.sql(CONSULTA_RETO).df()
display(resultado_reto)
print(f"Empresas que satisfacen ambos criterios: {len(resultado_reto)}")
```

**Pregunta 7.** Según los criterios establecidos, ¿qué empresa presenta la condición financiera más favorable dentro del grupo analizado?

**Respuesta :** Con los datos disponibles hasta ahora, solo Casa Grande satisface simultáneamente razón corriente ≥ 1 y ROE positivo entre las empresas confirmadas — Cartavio queda excluida por su razón corriente de 0.96. Esto no significa que Casa Grande sea "la mejor" en términos absolutos (su ROE es menor), sino que es la única, de las dos confirmadas, que cumple ambos criterios de riesgo definidos por la consulta. El resultado definitivo depende de incorporar Laredo y Pomalca.
